```python
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.regression import LinearRegression
 
print("Spark", spark.version)
df = spark.createDataFrame([(float(i), 2.0 * i + 1) for i in range(50)], ["x", "y"])
datos = VectorAssembler(inputCols=["x"], outputCol="features").transform(df)
modelo = LinearRegression(featuresCol="features", labelCol="y").fit(datos)
print("Pendiente:", round(modelo.coefficients[0], 2), "- Ordenada:", round(modelo.intercept, 2))
display(df)
```

```python
from delta.tables import DeltaTable
spark.range(5).write.format("delta").mode("overwrite").save("datos/delta_prueba")
display(spark.read.format("delta").load("datos/delta_prueba"))
display(DeltaTable.forPath(spark, "datos/delta_prueba").history())
```
